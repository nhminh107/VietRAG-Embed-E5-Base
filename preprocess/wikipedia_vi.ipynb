```python
import pandas as pd
from pathlib import Path

root_dir = Path.cwd().parent
data_dir = root_dir / "data" / "General"
wiki_1 = pd.read_parquet(data_dir / "wikipedia_vi_1.parquet")
wiki_2 = pd.read_parquet(data_dir / "wikipedia_vi_2.parquet")
wiki_3 = pd.read_parquet(data_dir / "wikipedia_vi_3.parquet")
```

```python
from database.sql_manager import SQL_Manager
from database.models import GeneralModel

id_prefix = "data_001"
source = "Wikipedia vi"
sql_mng = SQL_Manager()
sql_mng.create_general_model()
```

```python
def title_process(title: str):
    return f""""{title}" là gì?"""
```

```python
def chunk_content(content: str, chunk_size = 384, overlap = 48):
    content_chunk = []
    chunk_idx = 0
    while chunk_idx < len(content):
        end_idx = chunk_idx + chunk_size if chunk_idx + chunk_size < len(content) else len(content)
        content_chunk.append(content[chunk_idx:end_idx])
        chunk_idx = chunk_idx + chunk_size - overlap
    return content_chunk

print(chunk_content("""🚀 [GÓC KHOE PROJECT] TỔNG HỢP DỰ ÁN ĐANG THAM GIA CUỘC THI AI RISER VIETNAM  🚀
Đường đua AI Riser Vietnam 2026 đang bước vào giai đoạn nước rút! Chắc hẳn lúc này, các đội thi đang rất cần những người dùng thực tế trải nghiệm sản phẩm để thu thập feedback và hoàn thiện giải pháp của mình đúng honggg?
Hiểu được điều đó, GDG Cloud HCM dành riêng bài post này làm "trạm phát sóng" để hỗ trợ 100% công lực cho các dự án của anh em nhà mình:
 🌟 Tiếp cận user thực tế: Sản phẩm của bạn sẽ được giới thiệu trực tiếp đến các bạn thành viên đam mê công nghệ trong cộng đồng GDG Cloud HCM.
 💡 Nhận review "có tâm": Đội ngũ GDG Cloud HCM sẽ trực tiếp dùng thử, trải nghiệm và gửi lại cho bạn những feedback khách quan, hữu ích nhất (nếu có) để tối ưu sản phẩm trước khi nộp bài.
 🏆 Bứt tốc đua Top: Đừng quên, kêu gọi mọi người trải nghiệm và thả tim/reply comment chính là chiến thuật "nhỏ mà có võ" giúp dự án của bạn dễ dàng ẵm ngay giải thưởng "Top 3 bài có lượt tương tác cao nhất trên mạng xã hội" đấy nhé!
Đừng quên hạn cuối nộp bài: 23h59 ngày 30/8/2026. Và nhớ tick chọn GDG Cloud Ho Chi Minh City trong form nộp bài nha 💙🎉🔥"""))
```

```python
ver = 1
idx = 0
for index, row in wiki_1.iterrows():
    new_title = title_process(row['title'])
    contents = chunk_content(row['text'])
    for t in contents:
        gm = GeneralModel(
            data_id=f"{id_prefix}_{ver:03d}_{idx:010d}",
            source=source,
            title=row['title'],
            topic=None,
            anchor=new_title,
            positive=t,
            hard_negative=None
        )
        sql_mng.insert_general_model(gm)
        if idx % 1000 == 0:
            sql_mng.con.commit()
        idx += 1

        print(gm.data_id)

sql_mng.con.commit()
```
