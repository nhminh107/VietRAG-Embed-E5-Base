# PREPROCESS FOR VISP

```python
from operator import pos

import pandas as pd
from pathlib import Path

from dill import source
from numpy.ma.core import negative
from pandas.core.reshape import tile

root_dir = Path.cwd().parent
data_dir = root_dir / "data" / "General"
visp = pd.read_parquet(data_dir / "visp_train.parquet")

```

```python
from database.sql_manager import SQL_Manager
from database.models import GeneralModel

id_prefix = "data_002"
sql_mng = SQL_Manager()
```

```python
idx = 0
ver = 1

for index, row in visp.iterrows():
    gm = GeneralModel(
        data_id=f"{id_prefix}_{ver:03d}_{idx:010d}",
        source=row['source'],
        title=None,
        topic=row['topic'],
        anchor=row['original_text'],
        positive=row['paraphrase_text'],
        hard_negative=None
    )
    sql_mng.insert_general_model(gm)
    if idx % 1000 == 0:
        sql_mng.con.commit()
    idx+= 1

sql_mng.con.commit()
```
